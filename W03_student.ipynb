## q4: 5 marks
Write efficiency(out_w, in_w) returning the percentage, and returning None when in_w is zero or negative.

```python
# Your answer here
def efficiency(out_w, in_w):
    if in_w <= 0:
        return None
    else:
        return (out_w / in_w) * 100
```
